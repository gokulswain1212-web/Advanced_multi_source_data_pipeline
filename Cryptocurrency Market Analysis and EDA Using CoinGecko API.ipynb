```python
import requests
import pandas as pd

url = "https://api.coingecko.com/api/v3/coins/markets"

# Your API key
api_key = "CG-ikdsyYZQvQrg3cYtn3fReiY6"

headers = {
    "x-cg-demo-api-key": api_key
}

all_data = []

# Get 10 pages × 50 coins = 500 coins
for page in range(1, 11):

    params = {
        "vs_currency": "usd",
        "order": "market_cap_desc",
        "per_page": 50,
        "page": page
    }

    response = requests.get(
        url,
        params=params,
        headers=headers
    )

    print("Page:", page, "Status:", response.status_code)

    data = response.json()

    all_data.extend(data)

# Convert to DataFrame
df = pd.DataFrame(all_data)

print("Total rows:", len(df))
print(df.head())
```

```python
#1 Top 10 Cryptocurrencies by Market Cap — Matplotlib
top10 = df.nlargest(10, "Market_Cap")

plt.figure(figsize=(12, 6))

# Different color for each bar
colors = [
    "orange", "royalblue", "green", "gold", "purple",
    "red", "cyan", "pink", "brown", "teal"
]

plt.bar(
    top10["Name"],
    top10["Market_Cap"],
    color=colors
)

plt.title(
    "Top 10 Cryptocurrencies by Market Cap",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Cryptocurrency", fontsize=12)
plt.ylabel("Market Cap", fontsize=12)

plt.xticks(rotation=45)

# Add grid
plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```
